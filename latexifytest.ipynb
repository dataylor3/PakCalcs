```python
# 1. Define Symbols
x, y = sympy.symbols('x y')

# 2. Define Expression: E = x^2 + 5y
E = x**2 + 5 * y

# 3. Substitute values (x=3, y=4)
# The substitution is done using a dictionary mapping: {symbol: value}
E_substituted = E.subs({x: 3, y: 4}) 
print(f"Substituted Expression (Symbolic): {E_substituted}")

# 4. Evaluate to a numerical float
result = E_substituted.evalf()

print(f"Numerical Result (Float): {result}")
```

```python
# 1. Define Symbols
m, v, h = sympy.symbols('m v h')

# 2. Define Expression: Equation for Total Energy (E_total)
# E = 0.5 * m * v^2 + m * g * h
g = 9.81 # Define g as a constant value here for convenience
E_total = sympy.Rational(1, 2) * m * v**2 + m * g * h

# Display the original equation
print("Original Equation:")
display(Math(r"E_{\text{total}} = " + sympy.latex(E_total)))
```

```python
import sympy
from IPython.display import display, Math

# --- 1. SETUP: Define Symbols, Constants, and the Original Equation ---
# Define symbolic variables
M, V, H = sympy.symbols('M V H')

# Define the constant g (Gravity) and the values for substitution
g = 9.81
values = {M: 5, V: 10, H: 2}

# Define the equation for Total Energy (E_total): 0.5 * m * v^2 + m * g * h
E_total = sympy.Rational(1, 2) * M * V**2 + g * M * H

print("--- Original Symbolic Equation ---")
display(Math(r"E_{\text{total}} = " + sympy.latex(E_total)))

# --- 2. SUBSTITUTION: Replace Symbols with Numerical Values ---
# Use .subs() to replace the symbols. This result is the equation with numbers.
E_substituted = E_total.subs(values)

print("\n--- Equation with Substituted Values ---")
# Display the expression with the numbers substituted
display(Math(r"E_{\text{total}} = " + sympy.latex(E_substituted)))

# --- 3. EVALUATION: Calculate the Final Numerical Answer ---
# Use .evalf() to force the numerical calculation
final_result = E_substituted.evalf()

print("\n--- Final Numerical Result ---")
# Display the final numerical answer
display(Math(r"E_{\text{total}} = " + sympy.latex(final_result)))
```

```python
import sympy
from IPython.display import display, Math

# 1. Setup
M, V, H = sympy.symbols('M V H')
g = 9.81
values = {M: 5, V: 10, H: 2}
E_total = sympy.Rational(1, 2) * M * V**2 + g * M * H

# 2. SUBSTITUTION
E_substituted = E_total.subs(values)

print("--- Equation with Substituted Values (Forcing Display Method) ---")
# Use E_substituted._repr_latex_() directly inside display() to leverage SymPy's display logic
display(Math(r"E_{\text{total}} = " + E_substituted._repr_latex_()))

# 3. EVALUATION
E_result = E_substituted.evalf()

print("\n--- Final Numerical Result (Calculated) ---")
# Use the same method for the final result
display(Math(r"E_{\text{total}} = " + E_result._repr_latex_()))
```

```python
import sympy
from IPython.display import Math, display

# Example setup (assuming you have these defined elsewhere)
# Make sure these are SymPy symbols
E_kinetic, mass, velocity = sympy.symbols("E_kinetic mass velocity")
E_potential, spring_constant, displacement = sympy.symbols("E_potential spring_constant displacement")

# Define the expressions
E_kinetic_expr = (1/2) * mass * velocity**2
E_potential_expr = (1/2) * spring_constant * displacement**2
E_total = E_kinetic_expr + E_potential_expr

# Example substitutions (assuming you have these values)
mass_val = 2  # kg
velocity_val = 5  # m/s
spring_constant_val = 100  # N/m
displacement_val = 0.1  # m

# Create a dictionary for substitution
substitution_dict = {
    mass: mass_val,
    velocity: velocity_val,
    spring_constant: spring_constant_val,
    displacement: displacement_val,
}

# Substitute the values into the expression
E_substituted = E_total.subs(substitution_dict)

# Get the LaTeX representation of the substituted expression
latex_repr = sympy.latex(E_substituted)

# Display the LaTeX representation
display(Math(r"E_{\text{total}} = " + latex_repr))
```

```python
import sympy
from IPython.display import Math, display

# 1. Define the symbolic variables
# Using sympy.symbols is crucial for the library to track them.
E_total_sym, mass, velocity, k, x = sympy.symbols(r"E_{total} mass velocity k x")

# 2. Build the symbolic equation
# Use sympy.Rational for fractions to ensure they are treated symbolically
# and render nicely in LaTeX (e.g., as a fraction, not 0.5).
E_kinetic_expr = sympy.Rational(1, 2) * mass * velocity**2
E_potential_expr = sympy.Rational(1, 2) * k * x**2
E_total_expr = E_kinetic_expr + E_potential_expr

# 3. Define the numerical values for substitution
substitution_dict = {
    mass: 2,          # kg
    velocity: 5,      # m/s
    k: 100,           # N/m
    x: 0.1            # m
}

# 4. Perform the substitution WITHOUT evaluating the result
# This is the key change!
unevaluated_expr = E_total_expr.subs(substitution_dict, evaluate=False)

# 5. Generate the LaTeX string from the unevaluated expression
latex_str = sympy.latex(unevaluated_expr)

# 6. Display the final formatted output
display(Math(sympy.latex(E_total_sym) + " = " + latex_str))

# If you also want to show the final calculated value, you can do it separately
final_value = E_total_expr.subs(substitution_dict)
display(Math(sympy.latex(E_total_sym) + " = " + sympy.latex(final_value)))

```
